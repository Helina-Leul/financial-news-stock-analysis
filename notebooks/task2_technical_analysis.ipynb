```python
import talib

aapl["RSI"] = talib.RSI(
    aapl["Close"],
    timeperiod=14
)

aapl.head()
```

```python
plt.figure(figsize=(15,5))

plt.plot(
    aapl.index,
    aapl["RSI"]
)

plt.axhline(70)
plt.axhline(30)

plt.title("Apple RSI Indicator")
plt.xlabel("Date")
plt.ylabel("RSI")

plt.show()
```

The RSI indicator shows periods where Apple stock entered overbought or oversold conditions.
Values above 70 indicate strong buying pressure and possible price correction,
while values below 30 indicate possible undervaluation or selling pressure.

```python
macd, signal, hist = talib.MACD(
    aapl["Close"],
    fastperiod=12,
    slowperiod=26,
    signalperiod=9
)


aapl["MACD"] = macd
aapl["MACD_signal"] = signal
aapl["MACD_hist"] = hist


aapl.head()
```

```python
plt.figure(figsize=(15,5))

plt.plot(
    aapl.index,
    aapl["MACD"],
    label="MACD"
)

plt.plot(
    aapl.index,
    aapl["MACD_signal"],
    label="Signal"
)


plt.title("Apple MACD Indicator")

plt.xlabel("Date")
plt.ylabel("Value")

plt.legend()

plt.show()
```

MACD compares short-term and long-term price momentum.
When MACD crosses above the signal line it suggests increasing bullish momentum,
while crossing below indicates possible bearish momentum.

## Technical Indicator Summary

In this analysis, technical indicators were calculated to understand Apple stock price behavior.

The following indicators were implemented:

- SMA (Simple Moving Average): Identifies long-term and short-term price trends.
- EMA (Exponential Moving Average): Gives more importance to recent price movements.
- RSI (Relative Strength Index): Identifies overbought and oversold market conditions.
- MACD (Moving Average Convergence Divergence): Measures momentum changes and possible trend reversals.

These indicators help understand market trends and will later be combined with news sentiment analysis to study the relationship between financial news and stock movements.

```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from wordcloud import WordCloud

from sklearn.feature_extraction.text import CountVectorizer
from sklearn.feature_extraction.text import TfidfVectorizer

import warnings
warnings.filterwarnings("ignore")

plt.style.use("ggplot")
```

```python
news["headline_length"] = news["headline"].str.len()

news["headline_length"].describe()
```

```python
plt.figure(figsize=(10,5))

sns.histplot(
    news["headline_length"],
    bins=50,
    kde=True
)

plt.title("Distribution of Headline Lengths")
plt.xlabel("Headline Length")
plt.ylabel("Frequency")

plt.show()
```

```python
top_publishers = news["publisher"].value_counts().head(10)

top_publishers
```

```python
plt.figure(figsize=(12,6))

top_publishers.plot(kind="bar")

plt.title("Top 10 Publishers")
plt.xlabel("Publisher")
plt.ylabel("Number of Articles")

plt.xticks(rotation=45)

plt.show()
```

A small number of publishers contribute a large proportion of articles, indicating concentration in financial news production.

Top Keywords

```python
vectorizer = CountVectorizer(
    stop_words="english",
    max_features=20
)

X = vectorizer.fit_transform(news["headline"])

keywords = pd.DataFrame({
    "keyword": vectorizer.get_feature_names_out(),
    "count": X.toarray().sum(axis=0)
})

keywords = keywords.sort_values(
    by="count",
    ascending=False
)

keywords
```

```python
plt.figure(figsize=(12,6))

sns.barplot(
    data=keywords,
    x="count",
    y="keyword"
)

plt.title("Top Keywords in Financial Headlines")
plt.show()
```

## Task 1 Key Findings

### Dataset Overview
- The financial news dataset contains 1,407,328 articles.
- The dataset has 6 columns: headline, url, publisher, date, and stock symbol.
- No missing values were detected.

### Text Analysis
- Average headline length is around 73 characters.
- Most headlines focus on stock movements, earnings, estimates, market updates, and analyst opinions.

### Publisher Analysis
- The most active publishers are Paul Quintaro, Lisa Levin, and Benzinga Newsdesk.
- A small number of publishers contribute a large portion of articles.

### Time Series Analysis
- News volume changes significantly over time.
- Some periods show spikes in publication frequency, suggesting increased market activity.

### Keyword Analysis
- Common keywords include:
  - stocks
  - earnings
  - EPS
  - market
  - shares
  - price
  - trading

These themes show that the dataset mainly contains market-moving financial information.
